```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#style for figures
#sns.set_theme(font_scale=.75)
sns.set_style("whitegrid")
custom_colors={
    "control":"grey",
    "Takara PrimeSTAR LongSeq":"mediumpurple", 
    "NEB Q5 XT": "orange",
    "Kura Decodifi" : "skyblue"}
```

```python
df = pd.read_csv("../Data_files/sample_info_2026-07-31_with_features.csv")
```

```python
#splits dataframe into two dataframes for each mock community
eed=df.loc[df["origin"]=="EED fecal"]
esk=df.loc[df["origin"]=="ESK pathogen"]
```

```python
# remove all samples that went through Qiagen cleanup so as to not confound
eed_q=eed.loc[eed["qiagen_cleanup"]==False]
esk_q=esk.loc[esk["qiagen_cleanup"]==False]
```

```python
# remove all samples that have less than 100k reads
eed_nt=eed_q.loc[eed["reads_g100k"]==1]
esk_nt=esk_q.loc[esk["reads_g100k"]==1]
```

```python
#sorting samples to ensure they are plotted correctly with optimal colors
esk_nt_sort=esk_nt.sort_values(by="name_short_4", ascending=False)
eed_nt_sort=eed_nt.sort_values(by="name_short_4", ascending=False)
```

```python
# Richness ESK bactiera
plt.figure(figsize=(4,4))
ax=sns.barplot(data=esk_nt_sort, x="name_short_4", y="richness_bacteria_sylph_r232_observed_features", hue="mastermix", palette=custom_colors, legend=False)
ax.set(xlabel=None)
ax.set(ylabel="Bacterial diversity (Richness) ESK")
plt.xticks(rotation=45, ha="right")
plt.show
plt.savefig("../sylph-100k/results/ESK-bacterial_richness.png", dpi=300, bbox_inches='tight')
```

```python
# Richness ESK viral
plt.figure(figsize=(4,4))
ax=sns.barplot(data=esk_nt_sort, x="name_short_4", y="richness_viral_sylph_r232_observed_features", hue="mastermix", palette=custom_colors, legend=False)
ax.set(xlabel=None)
ax.set(ylabel="Viral diversity (Richness) ESK")
plt.xticks(rotation=45, ha="right")
plt.show
plt.savefig("../sylph-100k/results/ESK-viral_richness.png", dpi=300, bbox_inches='tight')
```

```python
# Richness ESK plasmid
plt.figure(figsize=(4,4))
ax=sns.barplot(data=esk_nt_sort, x="name_short_4", y="richness_n_plasmids", hue="mastermix", palette=custom_colors, legend=False)
ax.set(xlabel=None)
ax.set(ylabel="Plasmid diversity (Richness) ESK")
plt.xticks(rotation=45, ha="right")
plt.show
plt.savefig("../sylph-100k/results/ESK-plasmid_richness.png", dpi=300, bbox_inches='tight')
```

```python
# Richness EED bactiera
plt.figure(figsize=(5,4))
ax=sns.barplot(data=eed_nt_sort, x="name_short_4", y="richness_bacteria_sylph_r232_observed_features", hue="mastermix", palette=custom_colors, legend=False)
ax.set(xlabel=None)
ax.set(ylabel="Bacterial diversity (Richness) fecal")
plt.xticks(rotation=45, ha="right")
plt.show
plt.savefig("../sylph-100k/results/EED-bacterial_richness.png", dpi=300, bbox_inches='tight')
```

```python
# Richness EED viral
plt.figure(figsize=(5,4))
ax=sns.barplot(data=eed_nt_sort, x="name_short_4", y="richness_viral_sylph_r232_observed_features", hue="mastermix", palette=custom_colors, legend=False)
ax.set(xlabel=None)
ax.set(ylabel="Viral diversity (Richness) fecal")
plt.xticks(rotation=45, ha="right")
plt.show
plt.savefig("../sylph-100k/results/EED-viral_richness.png", dpi=300, bbox_inches='tight')
```
